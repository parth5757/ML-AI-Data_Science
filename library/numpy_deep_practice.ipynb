### Task 4
#### Indexing

```python
# Fancy Indexing

arr = np.array([10,20,30,40,50])
print(arr[[0,2,4]])

# Boolean Indexing
arr = np.array([5,10,15,20,25])
print(arr[arr > 15])
```

```python
# Modifying value using condition
arr = np.array([5,10,15,20])

if(arr[arr>15]):
  arr[arr<15] = 0
print(arr)
```

```python
# Copy V/S View
arr = np.array([1,2,3,4,5])
slice_part = arr[0:4]
slice_part[0] = 99
print(slice_part)


# copy arr
copy_arr = arr[1:4].copy()
print(copy_arr)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.plot([1,2,3,4])
plt.ylabel('some numbers')
plt.show()
```

```python
plt.plot([1, 2, 3, 5], [4, 9, 16, 10])
```

```python
x = [0,2,4,6,8]
y = [0,4,16,36,64]
fig, ax = plt.subplots()
ax.plot(x, y, marker='o', label="Data Points")
ax.set_title("Basoc component of Matplotlib")
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
```

#### linechart

```python
x = [10, 20, 30, 40]
y = [20, 25, 30, 35]
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

##### Bar Charts

```python
x = ['Thu', 'Fri', 'Sat', 'Sun']
y = [20, 25, 30, 35]
plt.bar(x, y)
plt.title("Bar Plot")
plt.xlabel("Days")
plt.ylabel("Values")
plt.show()
```

##### Histogram

```python
x = [7, 8, 9, 10, 10, 12, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20,
     21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40, 42, 44, 48, 50]

plt.hist(x, bins=10, color='steelblue')
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

##### Scatter Plot

```python
x = ['Thur', 'Fri', 'Sat', 'Sun', 'Thur', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190, 160, 130, 240, 200]

plt.scatter(x, y)
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

##### Pie Chart

```python
cars = ['AUDI', 'BMW', 'FORD','TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 12]

plt.pie(data, labels=cars)
plt.title("Pie Chart")
plt.show()
```

##### BoxPlot

```python
data = [ [10, 12, 14, 15, 18, 20, 22],
         [8, 9, 11, 13, 17, 19, 21],
         [14, 16, 18, 20, 23, 25, 27] ]

plt.boxplot(data)
plt.xlabel("Groups")
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

##### HeatMAP

```python
np.random.seed(0)
data = np.random.rand(10, 10)
plt.imshow(data, cmap='viridis', interpolation='nearest')

plt.colorbar()
plt.title("Heatmap")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## CountPlot|

```python
sns.countplot(df['Survived']) #Old Method now not work
```

```python
counts = df['Survived'].value_counts()

import matplotlib.pyplot as plt

plt.bar(counts.index.astype(str), counts.values)

plt.xlabel("Survived")
plt.ylabel("Count")
plt.title("Survival Count")

plt.show()
```

$$

$$

$$

```python
# Put Zero in unknown value
n1 = 1
d1 = 2
n2 = 0
d2 = 16

if n2==0:
  print (d2, n1/d1)
  ans = d2* n1 /d1
  print("n2 =", ans)

if d2==0:
  print(n2, d1/n1)
  ans = n2* d1 /n1
  print("d2 =", ans)
```

```python
import sympy
from sympy import symbols, Eq
from sympy.solvers import solve

x, y = symbols('x, y')

constraint = Eq(x, y)

# equation
eq = Eq(2*x + 3*y + 5, 50)

# a = solve(eq, constraint, x)
# b = solve(eq, constraint, y)
# print(a, b)

# Add an additional constraint (assuming x = y)
constraint = Eq(x, y)

# Solve the system of equations
solution = solve((eq, constraint), (x, y))

# Print the solution
print(f"x = {solution[x]}")
print(f"y = {solution[y]}")
```

Try to get value of x and y

```python
import sympy
from sympy import symbols, Eq
from sympy.solvers import solve

# Define the symbols
x, y = symbols('x y')

# Define the equation
eq = Eq(2*x + 3*y + 5, 50)

# Solve the equation for x and y
solution = solve((eq,), (x, y))

# Check if the solution is a list of tuples
if isinstance(solution, list) and len(solution) > 0 and isinstance(solution[0], tuple):
    x_value, y_value = solution[0]
    print(f"x = {x_value}")
    print(f"y = {y_value}")
else:
    print("Unexpected solution format:", solution)
```
